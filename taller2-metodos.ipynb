### **Taller 2: Transformación lineal en imágenes**

En esta notebook exploraremos cómo las transformaciones lineales pueden aplicarse a imágenes digitales.

Sea una imagen en escala de grises de dimensiones $h \times w$ píxeles. Podemos representarla como un vector columna

$$
\vec{x} \in \mathbb{R}^{d}, \qquad d = h \cdot w
$$

donde cada componente $x_i$ es el valor de intensidad del píxel $i$-ésimo, típicamente en el rango $[0, 255]$ o normalizado en $[0,1]$.

Así pues, una transformación lineal sobre la imagen consiste en aplicar una matriz $A \in \mathbb{R}^{d \times d}$ para obtener una nueva imagen:

$$
\vec{b} = A \cdot \vec{x}
$$

Importamos las librerías a utilizar:

$$

$$

#### **Transformación 2: Inversión de colores**

Para invertir los colores de la imagen (convertir blancos en negros y viceversa), multiplicamos cada píxel por $-1$ y luego sumamos $1$.

La matriz que realiza la primera parte de la operación es:

$$
A = -I = 
\begin{bmatrix}
-1 & 0 & 0 & \cdots & 0 \\
0 & -1 & 0 & \cdots & 0 \\
0 & 0 & -1 & \cdots & 0 \\
\vdots & \vdots & \vdots & \ddots & \vdots \\
0 & 0 & 0 & \cdots & -1
\end{bmatrix}
$$

donde $I$ es la matriz identidad de tamaño $d \cdot d$.

Sin embargo, la transformación lineal $A \cdot \vec{x}$ producirá valores negativos, y los píxeles deben estar normalizados en un rango de $[0, 1]$:

$$
A \cdot \vec{x} =
\begin{bmatrix}
-x_1 \\
-x_2 \\
\vdots \\
-x_d
\end{bmatrix}
$$

Para obtener la imagen invertida correctamente, necesitamos sumar $1$ a cada componente:

$$
\vec{b} = A \cdot \vec{x} + \vec{1} =
\begin{bmatrix}
1 - x_1 \\
1 - x_2 \\
\vdots \\
1 - x_d
\end{bmatrix}
$$
