1. ¿Por qué la reconstrucción tomográfica se considera un problema inverso? Identificar qué información se conoce, qué variable se busca recuperar y qué papel cumple el modelo $b = Ax$.

El modelo $b = Ax$ puede ser entendido como el efecto de una transformación lineal $T(x) \approx b$ donde $T:\mathbb{R}^{n²}\rightarrow \mathbb{R}^m$ sobre una tomografía $x$. Luego, decimos que la reconstrucción tomográfica se considera un problema inverso pues implica, a partir del $b \in \text{Codom}(T)$, obtener el $x \in \text{Dom(T)}$ que mejor aproxima dicha salida. En otras palabras, queremos partir del resultado o la salida para llegar a la entrada de la función.

2. ¿Cómo puede determinarse, a partir de $A$, si las mediciones permiten recuperar una única imagen? Relacionar la respuesta con el rango y el espacio nulo de $A$, y considerar qué ocurre cuando
existe $v \ne \bar{0}$ tal que $Av = \bar{0}$.

Sea $A$ una matriz de proyección $\in \mathbb{R}^{m\times n^2}$. Por lo visto anteriormente, sabemos que, si $rango(A) = n²$ (por teorema de rango y nulidad, es lo mismo que decir $dim(Nul\ A) = 0$), y asumiendo mediciones $b$ sin ruido, la imagen tomográfica siempre se puede reconstruir de manera única. Si existiera un $v \ne 0$ tal que $Av=0$, es decir, si $dim(Nul\ A) \ne 0$, sabemos que no necesariamente tendremos solución única para todas las tomografías.

3. ¿Cómo afecta el ruido al sistema de ecuaciones? Explicar qué problema resuelve el método de mínimos cuadrados, interpretar el residuo obtenido y discutir si su norma es suficiente para evaluar
la calidad de una reconstrucción.

Introducir ruido sobre los vectores de mediciones $b \in Im(T)$ puede dar como resultado nuevas mediciones perturbadas $b_\sigma$ que no necesariamente sigan perteneciendo a la imagen de la proyección. En otras palabras, el sistema $Ax=b_\sigma$ podría no tener solución. Dado este problema, lo que se desea es obtener una reconstrucción que se acerque tanto como sea posible a la tomografía $x$ que produjo las mediciones ruidosas. El método de mínimos cuadrados consiste en calcular la proyección otrogonal de $b_\sigma$ sobre el espacio columna de $A$, obteniendo como resultado un vector aproximado $\hat{b}$ que, por teorema, es la mejor aproximación que se puede obtener dentro de los $b \in Im(T)$.

El residuo que se obtiene de dicha operación funciona como una métrica que nos dice qué tanto nos alejamos de la medición ruidosa. Aún así, esta aproximación no es suficiente para evaluar su calidad como solución dado que podemos terminar por sobreajustar sobre el ruido al quedarnos con la mejor aproximación.

4. ¿Qué supuesto sobre la imagen introduce el término $||Dx||^2$? Analizar cómo cambian las reconstrucciones al modificar $\lambda$ y justificar las diferencias observadas.

El término $||Dx||²$ en la función objetivo a minimizar asume como supuesto que la medición con la que se trabaja no es exactamente la real. Por ello se introduce el término ya mencionado, pues evita terminar en soluciones, que justamente, sobreajusten el ruido en la medición. Aún así, este término no conoce la forma del ruido, simplemente desvía la mejor aproximación que podría obtenerse. De aquí, también surge el hiperparámetro $\lambda$ para poder controlar qué tan estricta es la penalización. En particular, se tiene que para un $\lambda$ chico la imagen reconstruída es nítida pero granular (producto del ruido) mientras que para un $\lambda$ grande la imagen logra suprimir la granularidad al coste de tener una imagen difusa o borrosa que no necesariamente conserva las estructuras originales de la imagen. Por ello, siempre se opta por un término medio y una búsqueda exhaustiva en pos de mediar en este *trade-off*.

5. Comparar `np.linalg.lstsq`, gradiente descendiente y Newton considerando las iteraciones necesarias, los parámetros que requiere cada método y las soluciones obtenidas. Explicar el comportamiento observado para Newton.

Por un lado, el método de mínimos cuadrados, a diferencia de los otros dos, siempre converge a la mejor aproximación (por teorema). Dado que se trata de un simple producto matricial, no requiere más computo que el calculo de $(A^TA)^{-1}$ (parte más demandante), sin embargo esta operación no es trivial a nivel computacional. Por ello, a veces podría ser necesario recurrir a metodos iterativos tales como gradiente descendiente y Newton.

Respecto a gradiente descendiente podemos decir que la cantidad de iteraciones depende del parámetro $\alpha$ (*learning rate*) así como de otros parámetros complementarios tales como `tolerancia` (diferencia entre la evaluación del paso actual y el anterior sobre la función objetivo) y `max_iter`(máxima cantidad de iteraciones). Si bien es el más liviano computacionalmente, requiere de una optimización previa de la tasa de aprendizaje para lograr algún tipo de convergencia, que además, suele terminar siendo un óptimo local. Más alla de eso, otro parámetro necesario para ejecutar el algoritmo es un punto inicial de evaluación, que puede ser aleatorio. En consecuencia, la reconstrucción obtenida mediante este método es la que parece presentar mayor residuo.

Finalmente, el método de Newton es similar al del gradiente descendiente en cuanto a parámetros solo que en lugar de emplear un parámetro $\alpha$ usa la inversa de la derivada segunda, es decir, el hessiano de la función objetivo como sustituto de la tasa de aprendizaje. De esta manera, el descenso aquí es más preciso al basar su descenso en una curva (Taylor de segundo orden) en lugar de una recta (Taylor de primer orden) como en el descenso del gradiente. Particularmente, para una función objetivo cuadrática se tiene que la iteración es una sola, puesto que la aproximación del Taylor resulta ser la propia función objetivo y presenta un Hessiano constante. De esta manera, se puede hallar el mínimo global directamente. Sin embargo, cuando la función es más compleja, este método se vuelve ineficiente pues el cálculo de la matriz hessiana es computacionalmente costoso.

6. Elaborar una síntesis que conecte el modelo $b = Ax$ con el rango, el espacio nulo, las proyecciones ortogonales, mínimos cuadrados y la regularización. Indicar qué aspecto del problema de
reconstrucción permite analizar cada concepto.

El modelo $b = Ax$ es una simplificación de la adquisición de imagenes tomograficas. Se parte desde un conjunto de mediciones producidas por los detectores y se quiere llegar a la imagen original que produjo dichos registros en las mediciones.

El teorema de rango y nulidad permite analizar y determinar la cantidad de soluciones posibles para cada conjunto de mediciones, es decir, la cantidad de imagenes que podrían haber producido un determinado registro de mediciones.

Sin embargo, cuando se introduce ruido en la obtención de las mediciones no siempre se puede llegar a una reconstrucción exacta incluso si la matriz de proyección produce rango-columna completo. Podría ser necesario hallar la mejor aproximación a las mediciones ruidosas dentro de nuestro conjunto de mediciones posibles. Es aquí cuando el método de mínimos cuadrados resulta de utilidad: calcula la proyección ortogonal de los registros ruidosos sobre el espacio columna de la matriz de proyección, que por teorema, es la mejor aproximación alcanzable.

A veces, el método de mínimos cuadrados termina por sobreajustar el ruido. Para ello se recurren a métodos iterativos tales como descenso del gradiente y Newton partiendo de una función objetivo que contemple la penalización por aproximarse demasiado al ruido. Esto se logra mediante el termino de regularización $\lambda||Dx||²$ donde según el parámetro $\lambda$ se elige el grado de penalización. Sin embargo, este método no sabe distinguir entre ruido real y la señal de las estructuras propias de la imagen. Por ello, es que resulta esencial la elección del parámetro de regularización.
